# Proposal-level occlusion assessment

This notebook introduces the depth-layer and voxel-density functions used in the FuDensityNet OR implementation. It uses a small synthetic example so that it can run without KITTI. The experimental evaluation uses actual RGB and LiDAR data through `scripts/evaluate_kitti.py`.


```python
import numpy as np
from fudensitynet.or_score import split_depth_layers_from_proposal, compute_voxel_density_for_points

depths = np.array([7.0, 7.2, 7.5, 8.0, 19.0, 19.5, 20.0, 20.1])
layers = split_depth_layers_from_proposal(depths, min_depth_gap=3.0)
print('Depth gap (m):', layers['max_depth_gap'])
print('Near / far points:', layers['n_near'], '/', layers['n_far'])
```

```python
near_points = np.array([[7.0,0.0,0.0],[7.1,0.1,0.0],[7.3,0.1,0.2],[7.6,0.0,0.1]], dtype=np.float32)
for size in (0.25, 0.50, 1.00):
    density, occupied, maximum = compute_voxel_density_for_points(near_points, size)
    print(f'{size:.2f} m: density={density:.3f}, occupied_voxels={int(occupied)}, max_count={int(maximum)}')
```

## Full KITTI workflow

1. Generate oracle proposals using the KITTI 2D box annotations for evaluation.
2. Calculate raw geometric descriptors from projected LiDAR points.
3. Fit percentile normalization and choose the OR threshold on calibration frames.
4. Evaluate OR ranking and threshold performance on held-out frames.

For a runnable end-to-end experiment, use the command in the repository README. This example does not fabricate KITTI detections or evaluation metrics.
